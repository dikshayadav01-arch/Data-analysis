```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\Diksha\Downloads\cust_satisfaction.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df["Customer Type"] = df["Customer Type"].str.strip().str.lower()
df["Type of Travel"] = df["Type of Travel"].str.strip().str.lower()
df["Gender"] = df["Gender"].str.strip().str.lower()
df["Class"] = df["Class"].str.strip().str.lower()
df["satisfaction"] = df["satisfaction"].str.strip().str.lower()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.columns
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df["gender"] = df["gender"].map({"male": 0, "female": 1})
df["satisfaction"] = df["satisfaction"].map({
    "neutral or dissatisfied": 0,
    "satisfied": 1
})
```

```python
df = pd.get_dummies(df, columns=["class", "type_of_travel"], drop_first=True)
```

```python
df.query("age > 20 and flight_distance > 200")
```

```python
df.sort_values(by="flight_distance", ascending=False)
```

```python
df.groupby("satisfaction")["age"].mean()
```

```python
df["gender"].value_counts()
```

```python
df["total_delay"] = df["departure_delay_in_minutes"] + df["arrival_delay_in_minutes"]
df.head()
```

```python
Q1 = df["flight_distance"].quantile(0.25)
Q3 = df["flight_distance"].quantile(0.75)
IQR = Q3 - Q1

df = df[(df["flight_distance"] >= Q1 - 1.5*IQR) &
        (df["flight_distance"] <= Q3 + 1.5*IQR)]
```

```python
df.corr(numeric_only=True)
```

```python
df.to_csv("cleaned_airline_data.csv", index=False)
```
