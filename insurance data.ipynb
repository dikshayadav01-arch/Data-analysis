```python
df.dropna(inplace=True)
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

no missing values , all columns are complete.

```python
df.groupby("smoker")["charges"].mean()
```

```python
sns.boxplot(x="smoker", y="charges", data=df)
plt.show()
```

smokers have high median or charges

```python
sns.scatterplot(x="age", y="charges", hue="smoker", data=df)
plt.show()
```

charges are increasing with age
Smoking, higher age, and higher BMI usually lead to higher medical charges.

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

1. heat map

   charges & smoker - Very strong

   charges & age - Moderate

   charges & bmi - Moderate

   charge & children - Weak

```python
df.groupby("region")["charges"].mean().plot(kind="bar")
plt.title("Average Insurance Charges by Region")
plt.xlabel("Region")
plt.ylabel("Average Charges")
plt.show()
```

Some regions have higher healthcare costs or risk factors.

```python
df.sort_values("age").plot(x="age", y="charges", kind="line")
plt.title("Age vs Charges")
plt.show()
```

Charges increase as age increases.

```python
df["age"].plot(kind="hist", bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()
```

How many people fall into each age range.

```python
df["smoker"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Smoker vs Non-Smoker Distribution")
plt.ylabel("")
plt.show()
```

**What data set tells**


how much each person is charged for insurance.

Smoking strongly increases insurance cost.

Insurance cost rises with age.

Higher BMI can mean higher health risk - higher insurance charges.

Different regions can have different medical costs or insurance policies.

This data helps predict how much insurance someone might pay based on their health, lifestyle, and demographics.
